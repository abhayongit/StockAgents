```python
import os, sys, anthropic, httpx

print(sys.executable, "| anthropic", anthropic.__version__, "| httpx", httpx.__version__)
for k in ["ANTHROPIC_BASE_URL", "HTTPS_PROXY", "HTTP_PROXY", "ALL_PROXY", "SSL_CERT_FILE", "REQUESTS_CA_BUNDLE"]:
    print(k, "=", os.environ.get(k))
print("key set:", bool(os.environ.get("ANTHROPIC_API_KEY")))

try:
    print("httpx ->", httpx.get("https://api.anthropic.com", timeout=15).status_code)
except Exception as e:
    print("httpx failed:", type(e).__name__, e, "| cause:", repr(e.__cause__))
```
